```python
import os

racine = "."

exclude = {".git", ".gitignore","Tristan_Files","9.EDA_V3","8.Autre_Modele","5.Preds","6.Data"}

for root, dirs, files in os.walk(racine):
    # supprimer les dossiers à exclure de la recherche
    dirs[:] = [d for d in dirs if d not in exclude]

    level = root.replace(racine, "").count(os.sep)
    indent = "    " * level
    print(f"{indent}{os.path.basename(root)}/")

    subindent = "    " * (level + 1)
    for f in files:
        print(f"{subindent}{f}")
```
