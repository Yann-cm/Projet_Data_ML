# Corrélation entre les features

## 1)Import des données

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_csv("../6.Data/kaggle_b2_fraud_train_v4.csv")
```

## 2) Observation

```python
import pandas as pd
import numpy as np

# =========================
# 1️⃣ Sélection des variables numériques
# =========================

df_numeric = df.select_dtypes(include=["int64", "float64"])

# =========================
# 2️⃣ Matrice de corrélation (valeur absolue)
# =========================

corr_matrix = df_numeric.corr().abs()

# =========================
# 3️⃣ On garde uniquement la partie supérieure
# (évite les doublons A,B et B,A)
# =========================

mask = np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
upper = corr_matrix.where(mask)

# =========================
# 4️⃣ On récupère les 20 plus fortes corrélations
# =========================

top_pairs = (
    upper.unstack()
    .dropna()
    .sort_values(ascending=False)
    .head(20)
)

# =========================
# 5️⃣ Affichage
# =========================

print("Top 20 des plus fortes corrélations :")
print(top_pairs)
```

```python
import pandas as pd
import numpy as np

# On garde uniquement les variables numériques
df_numeric = df.select_dtypes(include=["int64", "float64"])

# Calcul corrélations
corr_with_target = (
    df_numeric.corr()["target_is_fraud"]
    .drop("target_is_fraud")
)

# Tri par importance (valeur absolue)
corr_sorted = corr_with_target.reindex(
    corr_with_target.abs().sort_values(ascending=False).index
)

print(corr_sorted)
```

Les corrélations entre chargeback_resolution_time_days et post_event_status_code avec target_is_fraud sont trop corrélés. Ceci s'explique par le fait que ces deux colonnes ne prédisent pas la fraud, elles sont des conséquences de celle-ci. On supprime alors ces deux colonnes.

Pour les colonnes corrélés > 0,85, on les supprime pour une regression logistique.
Pour Random Forest on peut garder les colonnes qui sont < 0,95.

| colonne(s) concernée(s)                              | explication                                                                                           | action                                                                 |
|-------------------------------------------------------|--------------------------------------------------------------------------------------------------------|------------------------------------------------------------------------|
| chargeback_resolution_time_days                      | Variable post-événement : conséquence directe d’une fraude (data leakage)                            | Supprimer la colonne                                                  |
| post_event_status_code                               | Variable générée après investigation / traitement (data leakage)                                     | Supprimer la colonne                                                  |
| Variables avec corrélation > 0.85 (entre features)   | Forte multicolinéarité pouvant déstabiliser une régression logistique                                | Supprimer une des variables corrélées                                 |
| Variables avec corrélation > 0.95 (entre features)   | Corrélation quasi parfaite (information redondante)                                                  | Supprimer une des variables corrélées                                 |
| Variables avec corrélation < 0.95 (Random Forest)    | Les modèles basés sur les arbres gèrent mieux la multicolinéarité                                    | Conserver les variables                                               |
